```python
import os
import pickle
import numpy as np
from tensorflow.keras.preprocessing import image



# List of models and their names
models = [vgg16, vgg19, resnet, mobilenet, inception]
model_names = ['vgg16', 'vgg19', 'resnet', 'mobilenet', 'inception']

# Directory containing test images
test_path = '/content/drive/Othercomputers/My Laptop/RoadImagesDataSet/Test'

# Initialize lists
y_true = []
predictions = []

# Load previously saved predictions if available
try:
    with open('predictions.pkl', 'rb') as f:
        y_true, predictions = pickle.load(f)
except FileNotFoundError:
    pass

# Iterate through the images in the directory
for root, dirs, files in os.walk(test_path):
    for file in files:
        img_path = os.path.join(root, file)

        # Extract the original label
        remove = test_path + '/'
        original_label = root.replace(remove, "")
        y_true.append(original_label)

        # Load and preprocess the image
        img = image.load_img(img_path, target_size=(224, 224))  # Replace target size with your model's expected input size
        img_array = image.img_to_array(img)
        img_array = np.expand_dims(img_array, axis=0)

        # Iterate through the models and make predictions
        for model, model_name in zip(models, model_names):
            try:
                # Predict the class probabilities
                probabilities = model.predict(img_array)

                # Get the index of the class with the highest probability
                predicted_index = np.argmax(probabilities)

                # Append the prediction to the list
                predictions.append((f"{model_name}_ypred", predicted_index))
            except Exception as e:
                print(f"An error occurred with model {model_name}: {e}")

        # Save the current state of y_true and predictions
        with open('predictions.pkl', 'wb') as f:
            pickle.dump((y_true, predictions), f)

# Print the results
print(y_true)
print(predictions)
```
